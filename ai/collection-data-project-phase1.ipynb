```python
import os
import shutil

base_path = "/kaggle/input/datasets/srinivasbece/kindey-stone-dataset-splitted/Kindey_Stone_Dataset"

splits = ["train", "val", "test"]

classes = ["Cyst", "Normal", "Stone", "Tumor"]

output_base = "/kaggle/working/combined_dataset"

os.makedirs(output_base, exist_ok=True)
```

```python
for cls in classes:
    os.makedirs(os.path.join(output_base, cls), exist_ok=True)

print("Folders created successfully.")
```

```python
for split in splits:
    for cls in classes:
        source_folder = os.path.join(base_path, split, cls)
        target_folder = os.path.join(output_base, cls)

        if os.path.exists(source_folder):
            for file in os.listdir(source_folder):
                src_file = os.path.join(source_folder, file)
                
                new_name = f"{split}_{file}"
                dst_file = os.path.join(target_folder, new_name)

                shutil.copy(src_file, dst_file)

print("Images combined successfully.")
```

```python
zip_paths = []

for cls in classes:
    folder_path = os.path.join(output_base, cls)
    zip_path = f"/kaggle/working/{cls}.zip"
    
    shutil.make_archive(zip_path.replace(".zip", ""), 'zip', folder_path)
    zip_paths.append(zip_path)

print("Zipping completed.")
zip_paths
```

```python
all_zip_path = "/kaggle/working/all_classes.zip"
shutil.make_archive(all_zip_path.replace(".zip", ""), 'zip', output_base)

print("All dataset zipped.")
```
