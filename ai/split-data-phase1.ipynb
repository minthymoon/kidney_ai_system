```python
import os
import random
import shutil

input_base = "/kaggle/input/datasets/samerzaidan/dataset1"

output_base = "/kaggle/working/final_dataset"

classes = ["Cyst", "Normal", "Stone", "Tumor"]

splits = ["train", "val", "test"]

for split in splits:
    for cls in classes:
        os.makedirs(os.path.join(output_base, split, cls), exist_ok=True)

print("Folders ready.")
```

```python
TARGET_SIZE = 1370

balanced_data = {}

for cls in classes:
    class_path = os.path.join(input_base, cls)
    images = os.listdir(class_path)

    print(f"{cls} before: {len(images)}")

    # خلط عشوائي
    random.shuffle(images)

    # ناخد فقط 1370 صورة
    selected = images[:TARGET_SIZE]

    balanced_data[cls] = selected

    print(f"{cls} after: {len(selected)}")
```

```python
splits_ratio = {
    "train": 0.8,
    "val": 0.1,
    "test": 0.1
}

split_data = {}

for cls, images in balanced_data.items():
    total = len(images)

    train_end = int(total * 0.8)
    val_end = train_end + int(total * 0.1)

    split_data[cls] = {
        "train": images[:train_end],
        "val": images[train_end:val_end],
        "test": images[val_end:]
    }

    print(f"\n{cls}:")
    print(f"Train: {len(split_data[cls]['train'])}")
    print(f"Val: {len(split_data[cls]['val'])}")
    print(f"Test: {len(split_data[cls]['test'])}")
```

```python
for cls in classes:
    for split in splits:
        for img_name in split_data[cls][split]:
            
            src = os.path.join(input_base, cls, img_name)
            dst = os.path.join(output_base, split, cls, img_name)

            shutil.copy(src, dst)

print("Dataset split completed.")
```

```python
import shutil

final_dataset_path = "/kaggle/working/final_dataset"
zip_path = "/kaggle/working/final_dataset.zip"

shutil.make_archive(zip_path.replace(".zip", ""), 'zip', final_dataset_path)

print("Zipped successfully:", zip_path)
```
